### A - Coder une classe **Triangle** contenant 3 points **A**, **B** et **C**, un **constructeur** qui initialise ce triangle avec 3 points de coordonnées (0, 0), une méthode **perimetre()** qui renvoie la somme des longueurs des segments AB, BC et CA, et une méthode **barycentre()** qui renvoie un Point **G** avec les coordonnées du centre de gravité du triangle ABC.

image.png

```python
import math

class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def distance(self, other_point):
        return math.sqrt((self.x - other_point.x)**2 + (self.y - other_point.y)**2)

    def __str__(self):
        return f"({self.x}, {self.y})"

class Triangle:
    def __init__(self, pA=None, pB=None, pC=None):
        self.A = pA if pA is not None else Point(0, 0)
        self.B = pB if pB is not None else Point(0, 0)
        self.C = pC if pC is not None else Point(0, 0)

    def perimetre(self):
        side_ab = self.A.distance(self.B)
        side_bc = self.B.distance(self.C)
        side_ca = self.C.distance(self.A)
        return side_ab + side_bc + side_ca

    def barycentre(self):
        g_x = (self.A.x + self.B.x + self.C.x) / 3
        g_y = (self.A.y + self.B.y + self.C.y) / 3
        return Point(g_x, g_y)
```

### B - Tester la classe **Triangle** en créant 3 points **A**, **B** et **C**, puis en déplaçant A vers **(3, 2)**, B vers **(7, 5)**, et C vers **(5, 8)**, puis en affichant le  périmetre du triangle **ABC** et les coordonnées du point **G**, centre de gravité de ce triangle.

```python
A = Point(3, 2)
B = Point(7, 5)
C = Point(5, 8)

triangle_abc = Triangle(A, B, C)

perimeter = triangle_abc.perimetre()
barycenter_g = triangle_abc.barycentre()

print(f"Périmètre du triangle ABC: {perimeter}")
print(f"Coordonnées du barycentre G: ({barycenter_g.x}, {barycenter_g.y})")
```

### C - Afficher le triangle ABC et son centre de gravité dans un graphe en utilisant la bibliothèque MatPlotLib


```python
import matplotlib.pyplot as plt


abscisses = [A.x, B.x, C.x, A.x]
ordonnees = [A.y, B.y, C.y, A.y]


barycenter_x = barycenter_g.x
barycenter_y = barycenter_g.y

plt.figure(figsize=(8, 6))
plt.plot(abscisses, ordonnees, 'b-', label='Triangle ABC')
plt.plot([A.x, B.x, C.x], [A.y, B.y, C.y], 'ro', markersize=10, label='Vertices A, B, C')
plt.plot(barycenter_x, barycenter_y, 'gx', markersize=10, label='Barycenter G')

plt.annotate('A', (A.x, A.y), textcoords="offset points", xytext=(5,5), ha='center')
plt.annotate('B', (B.x, B.y), textcoords="offset points", xytext=(5,5), ha='center')
plt.annotate('C', (C.x, C.y), textcoords="offset points", xytext=(5,5), ha='center')
plt.annotate('G', (barycenter_x, barycenter_y), textcoords="offset points", xytext=(5,5), ha='center', color='green')

plt.title('Triangle ABC and its Barycenter G')
plt.xlabel('X-coordinate')
plt.ylabel('Y-coordinate')
plt.grid(True)
plt.axis('equal')
plt.legend()

plt.show()
```
