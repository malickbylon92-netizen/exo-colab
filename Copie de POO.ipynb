### Programmation Orientée Objet (POO) en Python

La programmation orientée objet (POO) permet de créer des entités (objets) que l'on peut manipuler.

Une classe regroupe des fonctions et des attributs qui définissent un objet. On appelle par ailleurs les fonctions d'une classe des " méthodes ".

Créons une classe Voiture avec 2 attributs, 4 méthodes et 1 constructeur :

```python
class Voiture:
  "Définition d'une voiture"
  # Attributs
  couleur = ""
  immatriculation = ""
  vitesse = 0
  # Méthodes
  def demarrer(self):
    self.demarree = True
    print("Moteur démarré")
    return
  def accelerer(self):
    if self.vitesse < 100:
     self.vitesse +=10
     print(self.vitesse)
    else:
     print("Vitesse limite atteinte")
    return
  def freiner(self):
    return
  def arreter(self):
    return
  # Constructeur
  def __init__(self):
    self.demarree = False
    print("Création d'un nouvel objet instance de la classe Voiture")
```

### Exercice :


*   Coder les méthodes **accelerer**() et **freiner**() pour une modification de la vitesse de **10 Km/h** à condition que l'attribut **demarree** soit égal à **True**.
*   Ces méthodes doivent afficher la nouvelle vitesse, qui ne pourra pas excéder 100 Km/h ni être négative.
*   Modifier le constructeur pour qu'on puisse passer en paramètres la **couleur** et l'**immatriculation**.
*   Coder un programme qui :
  - instancie une voiture **rouge** immatriculée **AB-123-CD** (constructeur),
  - démarre (demarrer()),
  - accélère 12 fois (accelerer() 12 fois),
  - s'arrête (arreter()),
  - freine 12 fois (freiner() 12 fois),
  - s'arrête (arreter()).



```python
class Voiture:
  "Définition d'une voiture"

  def __init__(self, couleur, immatriculation):
    self.couleur = couleur
    self.immatriculation = immatriculation
    self.demarree = False
    self.vitesse = 0
    print(f"Création d'une voiture {self.couleur} immatriculée {self.immatriculation}")

  def demarrer(self):
    if not self.demarree:
      self.demarree = True
      print("Moteur démarré")
    else:
      print("La voiture est déjà démarrée.")
    return

  def accelerer(self):
    if self.demarree:
      self.vitesse += 10
      if self.vitesse > 100:
        self.vitesse = 100
      print(f"Vitesse actuelle : {self.vitesse} Km/h")
    else:
      print("Impossible d'accélérer, la voiture n'est pas démarrée.")
    return

  def freiner(self):
    if self.demarree:
      self.vitesse -= 10
      if self.vitesse < 0:
        self.vitesse = 0
      print(f"Vitesse actuelle : {self.vitesse} Km/h")
    else:
      print("Impossible de freiner, la voiture n'est pas démarrée.")
    return

  def arreter(self):
    if self.demarree:
      self.demarree = False
      self.vitesse = 0
      print("Moteur arrêté. Vitesse : 0 Km/h")
    else:
      print("La voiture est déjà arrêtée.")
    return



print("\n--- Début du programme ---")


maVoiture = Voiture("rouge", "AB-123-CD")


maVoiture.demarrer()


for i in range(12):
  maVoiture.accelerer()


maVoiture.arreter()


print("\nTentative de freiner après l'arrêt :")
for i in range(12):
  maVoiture.freiner()


maVoiture.arreter()

print("\n--- Fin du programme  ---")
```
